# csv 데이터 불러오기

```python
import pandas as pd

car_names = ['쏘나타', '투싼', '팰리세이드', '스타리아', '싼타페','코나','스타렉스','제네시스','그랜저','아반떼']
dfs = []

for car in car_names:
    temp_df = pd.read_csv(f'data/{car}_기본_중복제거_차량번호_보험이력.csv')
    dfs.append(temp_df)

df = pd.concat(dfs, ignore_index=True)
df = df[df['가격'] > 100].copy()

# 계약중인 차량은 9999만원으로 표시하고 있으므로 데이터 제거

df = df[df['가격'] != 9999]

# 타차가해 또는 내차피해 금액이 0 이하인 이상치 데이터 제거

df = df[df['내차피해금액'] >= 0]
df = df[df['타차가해금액'] >= 0]
```

# 데이터 전처리 

### 차량 종류 칼럼 생성

```python
# 간단하게 차량의 모델 종류만 기입

car_names = ['그랜저', '아반떼', '쏘나타', '싼타페', '투싼', '팰리세이드', '스타렉스', '스타리아', '제네시스', '코나']


# 각 모델명이 포함된 경우 해당 모델명 할당
for car in car_names:
    df.loc[df['모델명'].str.contains(car, na=False), '모델 종류'] = car
    df.loc[df['제조사'].str.contains(car, na=False), '모델 종류'] = car
df['모델 종류'].value_counts()


# df.to_csv('현대_기본_중복제거_차량번호_보험이력.csv')
```

```python
# 차량 모델별 라벨 정의 (순서는 예시, 필요에 따라 수정 가능)
car_label_mapping = {
    '제네시스': 10,
    '팰리세이드': 9,
    '그랜저': 8,
    '싼타페': 7,
    '스타리아': 6,
    '쏘나타': 5,
    '투싼': 4,
    '코나': 3,
    '아반떼': 2,
    '스타렉스': 1 
}

# 각 모델명이 포함된 경우 해당 숫자 할당
for car, label in car_label_mapping.items():
    df.loc[df['모델명'].str.contains(car, na=False), '시리즈_명'] = label
    df.loc[df['제조사'].str.contains(car, na=False), '시리즈_명'] = label
df['시리즈_명'].value_counts()
```

### 데이터 라벨링

```python
import pandas as pd
import pickle
from sklearn.preprocessing import LabelEncoder

# ✅ 원본 복사
df1 = df.copy()

# ✅ 날짜 변환
df1["마지막 수정일"] = pd.to_datetime(df1["마지막 수정일"], format="%Y-%m-%d %H:%M:%S", errors='coerce')

# ✅ 범주형 컬럼 지정
label_cols = ["모델명", "배지", "세부 배지", "연료", "변속기", "등록 상태", "판매 유형", "판매지역", "판매처"]

# ✅ LabelEncoder 저장용 딕셔너리
label_encoders = {}

# ✅ 각 컬럼에 대해 라벨 인코딩 및 인코더 저장
for col in label_cols:
    df1[col] = df1[col].astype(str)  # NaN 방지를 위해 문자열로 변환
    le = LabelEncoder()
    df1[col] = le.fit_transform(df1[col])
    label_encoders[col] = le  # 인코더 저장

    # 🔒 각 인코더를 .pkl 파일로 저장
    with open(f'webmodel/{col}_encoder.pkl', 'wb') as f:
        pickle.dump(le, f)

# ✅ 결측치 보간
df1.ffill(inplace=True)

# ✅ 정렬
df1.sort_values(by="마지막 수정일", ascending=True, inplace=True)

# ✅ 필요 없는 컬럼 제거
df1 = df1.drop(columns='제조사').copy()

# ✅ 전처리 완료된 데이터 저장
df1.to_csv("data/현대차_전체_데이터(전처리완료).csv", index=False)

# 확인
df1.head()
```

# 칼럼 만들기

```python
from datetime import datetime

df1['총피해금액'] = df1['내차피해금액'] + df1['타차가해금액']
df1['총피해횟수'] = df1['내차피해'] + df1['타차가해']
df1['출시 경과'] = 2025 - df1['모델연도']
df1['침수'] = df1['전체침수횟수'] + df1['일부침수횟수']
df1["침수이력"] = df1["침수"].apply(lambda x: 1 if x > 0 else 0)


# '마지막 수정일'을 타임존이 없는 datetime으로 변환
df1["마지막 수정일"] = pd.to_datetime(df1["마지막 수정일"]).dt.tz_localize(None)

# 날짜 차이 계산 (등록경과일 생성)
df1["등록경과일"] = (datetime.now().replace(tzinfo=None) - df1["마지막 수정일"]).dt.days

```

# train, test 나누기
## 랜덤으로 선정된 500개의 데이터를 맞추는 방식

```python
from sklearn.model_selection import train_test_split

# 🎯 같은 랜덤 샘플이 유지되도록 random_state 고정
random_state = 1004  

# 🚀 학습 데이터(train)와 테스트 데이터(test) 분할
train, test = train_test_split(df1, test_size=500, random_state=random_state)

# 🔥 결과 확인
print(f"Train 데이터 개수: {len(train)}, Test 데이터 개수: {len(test)}")

# DataFrame을 CSV 파일로 저장할 때
train.to_csv('data/train.csv', index=False)  # train 데이터프레임을 'train.csv'로 저장
test.to_csv('data/test.csv', index=False)    # test 데이터프레임을 'test.csv'로 저장
```

###  가격 데이터 이상값 제거

```python
# 리스 및 렌트 차량 제거

train = train[~((train['모델 종류'] == '제네시스') & (train['가격'] <= 200))]
train = train[~((train['모델 종류'] == '아반떼') & (train['가격'] <= 700))]
train = train[~((train['모델 종류'] == '쏘나타') & (train['가격'] <= 700))]
train = train[~((train['모델 종류'] == '싼타페') & (train['가격'] <= 600))]
train = train[~((train['모델 종류'] == '팰리세이드') & (train['가격'] <= 900))]
train = train[~((train['모델 종류'] == '그랜저') & (train['가격'] <= 1000))]
```

```python
import pandas as pd

# 🔥 차종별 IQR 방식으로 가격 이상값 제거
df_clean = train.copy()  # 원본 유지

for car in car_names:
    Q1 = df_clean[df_clean['모델 종류'] == car]['가격'].quantile(0.25)  # 1사분위
    Q3 = df_clean[df_clean['모델 종류'] == car]['가격'].quantile(0.75)  # 3사분위
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR  # 하한선
    upper_bound = Q3 + 1.5 * IQR  # 상한선

    # 🚀 이상값 제거 (각 차종별 개별 적용)
    df_clean = df_clean[
        ~((df_clean['모델 종류'] == car) & 
          ((df_clean['가격'] < lower_bound) | (df_clean['가격'] > upper_bound)))
    ]

# ✅ 확인
print(df_clean['모델 종류'].value_counts())
print(df_clean[['모델 종류', '가격']].describe())


train = df_clean.copy()

df.to_csv('data/train_현대_기본_중복제거_차량번호_보험이력.csv')
```

## 상관계수 확인

1️⃣ 상관관계 히트맵이란?
상관관계 히트맵(Correlation Heatmap)은 변수 간의 관계를 시각적으로 표현하는 그래프임.

값의 범위는 -1부터 1까지로, 1에 가까울수록 양의 상관관계, -1에 가까울수록 음의 상관관계를 의미함.

0에 가까울수록 관계가 약함을 나타냄.

이번 히트맵에서는 |상관계수| 0.2 이상인 값만 표시하여, 의미 있는 변수 간의 관계만 확인할 수 있도록 필터링함.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 상관관계 행렬 계산
corr = train.drop(columns = ['차량번호','모델 종류']).corr()

# '가격'과의 절대 상관계수가 0.3 이상인 변수만 선택
high_corr_features = corr.index[abs(corr["가격"]) > 0.2]

# 선택된 변수만으로 상관계수 행렬 생성
filtered_corr = corr.loc[high_corr_features, high_corr_features]

# 히트맵 그리기
plt.figure(figsize=(10, 8))
sns.heatmap(filtered_corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Filtered Correlation Heatmap (|Corr| > 0.2)")
plt.show()

```

가격과 가장 높은 상관관계를 가진 변수는

🔸 모델명_평균가격(0.91)

🔸 모델_평균가격(0.66)

→ 모델명과 모델에 따른 평균 가격이 예측에 중요한 피처임을 보여줍니다.

총피해횟수, 내차피해, 타차가해
서로 간 상관성이 높고 (0.8~0.9 수준),
**가격과는 약한 상관**을 보입니다.

→ 피해 관련 데이터는 가격에 어느 정도 영향을 주지만, 주요 변수는 아님을 시사합니다.

출시경과, 연식, 모델연도 간 음의 상관관계가 뚜렷합니다.

→ 출시 경과가 오래될수록 연식이 낮고 가격도 낮아지는 패턴을 설명합니다.

주행거리는 예상대로 가격과 음의 상관을 가지고 있음 (-0.50)

→ 많이 탄 차량일수록 가격이 낮아지는 일반적인 중고차 시장 흐름을 반영합니다.

## CatBoost 모델

CatBoost 모델의 성능을 최적화하기 위해 Optuna를 활용하여 자동으로 최적의 하이퍼파라미터를 찾았습니다.

Optuna는 단순한 랜덤 탐색이 아니라, 베이지안 최적화 방식을 활용하여 학습 과정에서 성능이 좋은 방향으로 빠르게 수렴하도록 도와줍니다.

또한 5-Fold 교차 검증을 통해 검증 데이터를 활용했으며, NMAE 값을 줄이는 방향으로 모델을 튜닝했습니다.

Optuna를 사용함으로써 실험 시간을 단축하면서도 최적의 하이퍼파라미터를 찾아내어 모델 성능을 극대화할 수 있었습니다.

```python
import optuna
import catboost as cb
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
import numpy as np

# 🎯 NMAE 함수 정의
def nmae(true, pred):
    return np.mean(np.abs(true - pred)) / np.mean(true)

# ✅ Train / Validation 분리
X_tr, X_val, y_tr, y_val = train_test_split(X_train, Y_train, test_size=0.2, random_state=1004)

# 🎯 Optuna 목적 함수 (Validation 기준 NMAE 최적화)
def objective(trial):
    params = {
        "depth": trial.suggest_int("depth", 3, 12),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.1, log=True),
        "iterations": trial.suggest_int("iterations", 100, 1000, step=50),
        "l2_leaf_reg": trial.suggest_float("l2_leaf_reg", 1e-3, 10.0, log=True),
        "border_count": trial.suggest_int("border_count", 32, 255),
        "subsample": trial.suggest_float("subsample", 0.7, 1.0),
        "random_seed": 1004,
        "loss_function": "MAE",
        "task_type": "CPU",
        "verbose": 0,
    }

    model = cb.CatBoostRegressor(**params)
    model.fit(X_tr, y_tr)

    preds = model.predict(X_val)
    score = nmae(y_val, preds)
    return score  # NMAE가 작을수록 좋음

# ✅ Optuna 최적화 시작
study = optuna.create_study(direction="minimize", sampler=optuna.samplers.TPESampler(seed=1004))
study.optimize(objective, n_trials=50)

# ✅ 최적 하이퍼파라미터 출력
print("✅ Best NMAE:", study.best_value)
print("✅ Best Hyperparameters:", study.best_params)
```
